```python
import torch
from torchvision.datasets import OxfordIIITPet
from torchvision import transforms
from torch.utils.data import DataLoader, Subset
```

```python
transform = transforms.Compose([
    transforms.Lambda(lambda img: img.convert("RGB")),
    transforms.Resize((128, 128)),
    transforms.ToTensor()
])

# training and validation
dev_dataset = OxfordIIITPet(
    root="/content/data",
    split="trainval",
    target_types="category",
    transform=transform,
    download=True
)

# test set
test_dataset = OxfordIIITPet(
    root="/content/data",
    split="test",
    target_types="category",
    transform=transform,
    download=True
)

print("Development images:", len(dev_dataset))
print("Test images:", len(test_dataset))
```

```python
train_size = int(0.8 * len(dev_dataset))
val_size = len(dev_dataset) - train_size

generator = torch.Generator().manual_seed(120391209371)

train_subset, val_subset = torch.utils.data.random_split(
    dev_dataset,
    [train_size, val_size],
    generator=generator
)

print("Training:", len(train_subset))
print("Validation:", len(val_subset))
print("Test:", len(test_dataset))
```

```python
split_data = {
    "train_indices": train_subset.indices,
    "val_indices": val_subset.indices
}

torch.save(split_data, "/content/data_split.pt")

print("Split saved.")
```
